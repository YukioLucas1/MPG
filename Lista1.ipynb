```python
import math
r = float(input())
print("{:.4f}".format(math.pi * r ** 2))
```

```python
import math
a = float(input())
b = float(input())
c = float(input())

valor_de_V = "{:.2f}".format(a * b * c)
valor_de_d = "{:.2f}".format(math.sqrt(a**2 + b**2 + c**2))

print("Volume:", valor_de_V, "Diagonal:", valor_de_d)
```

```python
import math
a = int(input())
b = int(input())
x = float(input())
y = float(input())

calculo = "{:.2f}".format(a + b**x - math.sqrt(b) + ((a + b)/(x - y)))
print("expressao: ", calculo)
```

```python
xUm = float(input())
yUm = float(input())
xDois = float(input())
yDois = float(input())

space = "{:.4f}".format(abs(xUm - xDois) + abs(yUm - yDois))
print(space)
```

```python
import math
a = float(input())
b = float(input())
c = float(input())
x = float(input())

Delta = b**2 -4*a*c
RaizUm = "{:.2f}".format((-1*b + math.sqrt(Delta))/(2*a))
RaizDois = "{:.2f}".format((-1*b - math.sqrt(Delta))/(2*a))
valor = "{:.2f}".format(a*x**2 +b*x+c)

print("Raizes: ", RaizUm, RaizDois, "Resultado para x = ", x,":", valor)
```
